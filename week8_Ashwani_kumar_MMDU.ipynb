## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

### 🚀 Bonus

````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
# 🤖 AGENT FUNCTION (COMPLETED)

def agent(query: str):
    query_lower = query.lower()

    try:
        # Calculator Routing
        if "calculate" in query_lower:
            expression = query_lower.replace("calculate", "").strip()

            return {
                "type": "calculation",
                "result": calculator(expression)
            }

        # Keyword Extraction Routing
        elif "keyword" in query_lower:
            text = query
            text = text.replace("Extract keywords from", "")
            text = text.replace("extract keywords from", "")
            text = text.strip()

            return {
                "type": "keywords",
                "result": extract_keywords(text)
            }

        # General Query
        else:
            return {
                "type": "general",
                "result": "This is a general query. No tool was required."
            }

    except Exception as e:
        return {
            "type": "error",
            "result": str(e)
        }
````

````python
# 🔄 Agent Pipeline Diagram
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(12, 3))
ax.axis('off')
ax.set_xlim(0, 10)
ax.set_ylim(0, 3)

steps  = ["👤 User\nQuery", "🤖 Agent\nRouter", "🔢 Calculator\nTool", "📝 Keyword\nExtractor", "💬 General\nResponse", "✅ Structured\nJSON Output"]
colors = ['#AED6F1','#D7BDE2','#A9DFBF','#A9DFBF','#A9DFBF','#82E0AA']
x_pos  = [0.3, 2.0, 3.7, 5.4, 7.1, 8.8]

for i, (s, c, x) in enumerate(zip(steps, colors, x_pos)):
    ax.add_patch(plt.Rectangle((x, 0.8), 1.5, 1.2, color=c, ec='gray', lw=1.5))
    ax.text(x+0.75, 1.4, s, ha='center', va='center', fontsize=8, fontweight='bold')
    if i < len(steps)-1:
        ax.annotate('', xy=(x_pos[i+1], 1.4), xytext=(x+1.5, 1.4),
                    arrowprops=dict(arrowstyle='->', color='black', lw=2))

ax.set_title('Single Agent Pipeline — Conditional Routing Architecture',
             fontsize=11, fontweight='bold', pad=10)
plt.tight_layout()
plt.show()
print("✅ Pipeline diagram complete")
````
